```python
#необходимо заменить значения "ремонта" на числовые.

df = pd.read_excel('Линейная регрессия.xlsx')
print("Уникальные значения до замены:")
print(df['repair'].unique())

df['repair'] = df['repair'].astype(str).str.strip().str.lower()

condition_map = {
    'none': 0,
    'bad': 1,
    'good': 2,
    'perfect': 3
}
df['repair'] = df['repair'].map(condition_map)

print("\nПосле замены:")
print(df['repair'].value_counts())
print("\nПервые 10 строк:")
print(df[['repair']].head(10))
```

```python
correlations = df.corr()['price'].sort_values(ascending=False)
print(correlations)
df.hist(figsize=(15, 10), bins=20, edgecolor='black')
plt.suptitle('Распределение признаков')
plt.show()
```

```python
#видим, что балкон, этажность и состояние ремонта имеют низкую корреляцию. Для линейной регрессии это не очень хорошо, для случайного леса - не существенно. Для сравнения я не буду удалять эти признаки.
```

```python
x = df.drop('price', axis=1)
y = df['price'] #целевая переменная - y. предсказывается по х.

from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42) #разделили на обучающую и тестовую выборку 80\20

```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train, y_train)
lr_pred = lr.predict(x_test)
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
print("=== Linear Regression ===")
print(f"MAE: {mean_absolute_error(y_test, lr_pred):.0f}")
print(f"RMSE: {np.sqrt(mean_squared_error(y_test, lr_pred)):.0f}")
print(f"R2: {r2_score(y_test, lr_pred):.3f}")
```

```python
#Вывод: модель работает. 23% вариации цены модель не уловила — нужны более качественные признаки или нелинейные зависимости.
plt.scatter(y_test, lr_pred, alpha=0.5)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2)
plt.xlabel('Реальная цена')
plt.ylabel('Предсказанная цена')
plt.title('Линейная регрессия: реальные vs предсказанные цены')
plt.show() 

residuals = y_test - lr_pred
plt.hist(residuals, bins=20, edgecolor='black')
plt.title('Распределение ошибок линейной регрессии')
plt.xlabel('Ошибка (реальная - предсказанная)')
plt.ylabel('Частота')
plt.show() #график ошибок
```

```python
df_rf = pd.read_excel('Random Forest.xlsx')

df_rf['repair'] = df_rf['repair'].astype(str).str.strip().str.lower()

condition_map2 = {
    'none': 0,
    'bad': 1,
    'good': 2,
    'perfect': 3
}
df_rf['repair'] = df_rf['repair'].map(condition_map2)
```

```python
X_rf = df_rf[['total_area', 'living_area', 'number_of_rooms', 'building_year', 'repair', 'distance_park_km', 'balcony', 'floor', 'total_floor']]
y_rf = df_rf['price']
X_rf = df_rf.drop('price', axis=1)
y_rf = df_rf['price']

from sklearn.model_selection import train_test_split
X_train_rf, X_test_rf, y_train_rf, y_test_rf = train_test_split(X_rf, y_rf, test_size=0.2, random_state=42)

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

rf = RandomForestRegressor(n_estimators=100, random_state=42)
rf.fit(X_train_rf, y_train_rf)
rf_pred = rf.predict(X_test_rf)

print("=== Random Forest (полные данные) ===")
print(f"MAE: {mean_absolute_error(y_test_rf, rf_pred):.0f}")
print(f"RMSE: {np.sqrt(mean_squared_error(y_test_rf, rf_pred)):.0f}")
print(f"R2: {r2_score(y_test_rf, rf_pred):.3f}")
```

```python
plt.scatter(y_test_rf, rf_pred, alpha=0.5)
plt.plot([y_test_rf.min(), y_test_rf.max()], [y_test_rf.min(), y_test_rf.max()], 'r--', lw=2)
plt.xlabel('Реальная цена')
plt.ylabel('Предсказанная цена')
plt.title('Random Forest: реальные vs предсказанные цены')
plt.show()

residuals_rf = y_test_rf - rf_pred
plt.hist(residuals_rf, bins=20, edgecolor='black')
plt.title('Распределение ошибок Random Forest')
plt.xlabel('Ошибка (реальная - предсказанная)')
plt.ylabel('Частота')
plt.show()
```

```python
#проверим модель на новой квартире, не принадлежащей выборке
new_flat = pd.DataFrame([[
    56.6,   # total_area
    31.0,   # living_area
    2,      # number_of_rooms
    1,      # balcony (1 - есть)
    2,      # repair (0=none,1=bad,2=good,3=perfect)
    5,      # floor
    6,      # total_floor
    2018,   # building_year
    0.8     # distance_park_km
]], columns=['total_area', 'living_area', 'number_of_rooms', 'balcony', 'repair', 'floor', 'total_floor', 'building_year', 'distance_park_km'])

predicted_price = rf.predict(new_flat)
print(f"Предсказанная цена: {predicted_price[0]:.0f} рублей")
```
